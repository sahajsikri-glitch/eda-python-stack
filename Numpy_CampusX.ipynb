# **ITERATION** **IN** **NUMPY** **ARRAYS**

```python
a1=np.arange(1,11)
a2=np.arange(1,11).reshape(2,5)
a3=np.arange(1,28).reshape(3,3,3)
print(a1,"\n")
print(a2,'\n')
print(a3,'\n')

for i in a1:#1d
  print(i)
print('\n')
for i in a2:#2d
  print(i) # printing rows
print('\n')
for i in a3:#3d
  print(i) # printing 2d arrays
print('\n')

#solution : convert every D into 1D :
for i in np.nditer(a3):
  print(i)
print('\n')
```

# **STACKING** AN**D** **SPLITTING**

```python
#IMP : shape should be SAME to stack 2 arrays
a0=np.arange(1,13).reshape(3,4)
a00=np.arange(13,25).reshape(3,4)
print(a0)
print(a00)

#HORIZONTAL STACKING :
ah= np.hstack((a0,a00))
print(ah)

#Vertical stacking :
av = np.vstack((a0,a00))
print(av)
```

```python
#HORIZONTAL SPLITTING :-> vertical line kaata
#VERTICAL SPLITTING :-> horizontal line kaata
print(np.hsplit(a0,4))
# np.hsplit(0,5) WRONG -> not equal division
print(np.vsplit(a0,3))
```

# np.corrcoef -> -1 to 1

```python
# -1 -> inversly propotional ek kam toh dusra bada
# 0 -> not related ek bada toh dusre ko laand ni farak padta
# 1 -> propotional jitna bada utna vo bada
```

```python
salary = np.array([2000,1000,3000,8000,2880])
experience = np.array([1,2,3,1,2])
np.corrcoef(salary,experience)
# format of output :
# [salary ka salary se=obv 1 | exp ka salary se
# exp ka salary se    | salary ka salary se=obv 1]
```
